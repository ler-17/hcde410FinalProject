# Overview, Background, and Research questions
The Gutenberg Poetry Corpus contains over three million lines of public-domain poetry from Project Gutenberg. It was assembled to support computational poetry generation, i.e. the use of language models to generate poetry or explore poetic ideas [1]. Language models tend to reproduce, and sometimes amplify, the biases of their training data, and language data is no exception. Models often handle underrepresented dialects poorly, misreading dialect input or producing stereotyped imitations [2, 3]. Because the corpus is assembled from published sources (i.e. more likely to represent "standard" langauge varities) and historical, it may favor some English varieties and underrepresent others, a skew that generators trained on it could inherit. This project examines the corpus’s langauge composition through two questions:

- How are dialect-marked and archaic lexical features distributed across the Gutenberg Poetry Corpus, and how does this vary by era?
- What do these distributions suggest about which English varieties a generator trained on it would reproduce or omit?

# Methodology 
I will conduct all analysis in a single Python Jupyter notebook with fixed random seeds and local copies each dataset, so every table and figure can be reproduced by rerunning it.

**Datasets:** I will be using two datasets for this project. 
- (1) The Gutenberg Poetry corpus itself (https://github.com/aparrish/gutenberg-poetry-corpus) which I will analyze for lexical and orthographic features (more detail below) and
- (2) the Gutenberg, Dammit dataset (https://github.com/aparrish/gutenberg-dammit/) which includes important metadata from Project Gutenberg, used for a period and broad language variety analysis (more detail below)

**Lexical / Orthographic feature prevalence:** I will build a small marker list in three categories: archaic (e.g., thou, o’er), Scots/regional (e.g., dinna, bairn), and elision/eye-dialect (e.g., singin’). Each marker will cite a published (and preferably peer-reveiwed) source. I will count matches per 1,000 lines by era. Frequency counts will be transparent, replicable. A full dialect analysis would capture more than orthographic and lexical features, but for the sake of meeting the project deadline, these features will be sufficient in providing a general idea of the corpus's dialect makeup. 

**Validation/Manual Check** I will hand-label about 100 randomly sampled matches per category (archaic, regional, eye-diact). Time could constrain the sample size, but the point of this is to check the precision of my analysis against manual coding, and demonstrate how noise could affect the counts.

**Macro-Composition analysis** I will join each line in the Poetry Corpus to its Gutenberg, Dammit metadata by Gutenberg ID, then report the share of lines by author birth/death era and language tag, along with metadata coverage. This shows which macro-periods the corpus emphasizes, and the coverage check shows how far the findings generalize. 

**Presentation** Findings will be presented in a Python Notebook which will show all data analysis, results, and interpretation. Implications for language models will be framed as hypotheses rather than demonstrated outcomes, since I will not have time to test the findings on existing models. Additionlly, a short presentation of my findings and methods will be prepared to share with a small group of peers. Here, I will practice my research presentation and receive feedback.
